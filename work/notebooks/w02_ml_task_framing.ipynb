## Why ML beats a fixed rule here

A fixed rule can be useful as a baseline. For example, a rule could prioritize pages that have not been updated for 180 days and still receive a high number of impressions.

However, a fixed rule depends on manually selected thresholds and may miss interactions between multiple signals.

Machine learning can learn patterns across several observable features, such as content age, days since the last update, impressions, average position, CTR, and word count.

The purpose is not to assume that ML will always outperform a human rule. I would test whether the model actually improves the ranking quality using a suitable evaluation method.

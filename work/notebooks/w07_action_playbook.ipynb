## 1. Ranked actions + reason codes

### Purpose

The model is used to rank content pages for **human review**. The ranking is not an automatic decision that a page must be refreshed.

The validated Decision Tree achieved a **Precision@50 of 0.620** on the client-holdout test set. I therefore use the model output as a prioritization signal rather than as a guaranteed prediction.

### Ranked action framework

Each recommended page should receive:

1. A model score or ranking position.
2. A reason code explaining why it was prioritized.
3. A suggested human-review action.
4. A clear indication that the recommendation requires human validation.

### Reason codes

| Reason code | Signal | Suggested action |
|---|---|---|
| `LOW_VISIBILITY` | Low recent impressions | Check whether the page has sufficient search visibility before investing in a refresh |
| `AGING_CONTENT` | High content age | Review whether information, examples, or structure are outdated |
| `LOW_CTR` | Low CTR relative to the available visibility signal | Review title/meta description and search-intent alignment |
| `LOW_POSITION` | Weak average search position | Review search intent, topical coverage, internal linking, and content usefulness |
| `REFRESH_CANDIDATE` | Multiple review signals occur together | Prioritize for a deeper human content audit |
| `REVIEW_ONLY` | Model ranking provides a signal but no single clear action | Require manual investigation before taking action |

These reason codes are decision-support labels. They do not establish that any individual feature caused decline.

### Recommended action priority

The highest-ranked pages should be reviewed first. Human reviewers should consider the model ranking together with business value, search intent, content quality, freshness, and the cost of making changes.

The model should therefore answer:

> "Which pages should we investigate first?"

rather than:

> "Which pages must be changed?"

## 3. Human review + the no-go list

### Human review workflow

The ranked queue is a prioritization aid. Before any content action is taken, a human reviewer should:

1. Verify the page and its current business importance.
2. Check whether the observed decline is meaningful and whether the page has enough recent visibility.
3. Review search intent, content quality, freshness, and topical coverage.
4. Check for possible technical issues, cannibalization, or other explanations for the observed signal.
5. Consider the expected value of the change versus the effort required.
6. Choose one of four outcomes:
   - **Refresh**
   - **Monitor**
   - **No action**
   - **Further investigation**

### Archetype → action mapping

| Content archetype | Review focus | Suggested action |
|---|---|---|
| Young + Fresh | Usually lower refresh priority | Monitor unless other signals are concerning |
| Mid-age + Stale | Check whether performance has started to decay | Review for possible refresh |
| Old + Refreshed | Check whether previous refreshes remain effective | Monitor and maintain when performance is healthy |
| Old + Long-Unchanged | Check freshness, relevance, and business value carefully | Prioritize for human review, but do not assume refresh is required |

These archetypes describe observed portfolio patterns. They are not automatic rules for individual pages.

### Decay and refresh insight

The portfolio analysis suggests that content performance can vary with age and freshness. In the research dataset, health was strongest in the **61–90 day** age band and lower in the **271–365 day** band. Older content that had been refreshed recently also showed stronger observed health than comparable stale content.

This supports using **content age and freshness as review signals**, but it does not prove that refreshing a page will cause its performance to improve.

### No-go list

The following actions should **not** be automated from this model:

- Automatically publishing rewritten content.
- Automatically changing titles or meta descriptions.
- Automatically deleting or redirecting pages.
- Automatically declaring a page low-quality.
- Automatically changing a page's business priority.
- Automatically treating feature importance as causal evidence.
- Automatically refreshing every high-ranked page.
- Automatically reporting the model score as a guaranteed prediction.

The model output must remain a **human-reviewed decision-support queue**.

## 4. Monitoring / retrain triggers

### Monitoring approach

The model should be monitored as a decision-support system rather than treated as a fixed production predictor.

The following signals should be checked periodically:

| Monitoring signal | Why it matters | Suggested response |
|---|---|---|
| Precision@50 | Measures how useful the top-ranked review queue remains | Re-evaluate the model if performance falls materially |
| Declining-rate shift | Changes in the target distribution can affect ranking usefulness | Investigate the underlying data before relying on the queue |
| Feature distribution shift | New client or time-period patterns may differ from training data | Review whether the model still represents the current data |
| Client mix changes | The model was validated using client-holdout evaluation | Re-test when substantially different clients are introduced |
| Review outcomes | Human reviewers provide evidence about whether recommendations are useful | Use accumulated outcomes to assess future model updates |

### Retrain / revalidation triggers

A new model should not be trained simply because more data becomes available. Retraining or revalidation should be considered when:

- Precision@50 declines materially on a new validation period.
- The distribution of important input features changes substantially.
- The client mix becomes materially different from the training data.
- The definition of the target or available source data changes.
- Human reviewers consistently find that the ranked queue is no longer useful.

Any retrained model should again be evaluated using an appropriate client-level or otherwise leakage-resistant validation design before being used for decision support.

### Monitoring principle

The queue should be treated as **revisable decision support**, not a permanently valid production rule.

```python
# Section 5: Export the ranked content-review queue.

import os
import pandas as pd
from sklearn.model_selection import GroupShuffleSplit
from sklearn.tree import DecisionTreeClassifier

# Final model features used in W05
features = [
    "content_age_days",
    "days_since_last_update",
    "impressions_90d",
    "avg_position",
    "ctr",
    "word_count"
]

# Load the dataset
data_path = "/content/Flyrank_Internship_1/data/raw/content_refresh_anonymized.csv"
df = pd.read_csv(data_path)

print("Dataset loaded:", df.shape)

# Create the target
df["is_declining_label"] = (
    df["trend_direction"].str.lower().eq("down").astype(int)
)

# Keep rows needed for modeling
model_data = df.dropna(
    subset=features + ["client_id", "is_declining_label"]
).copy()

# Recreate the validated client-holdout split
splitter = GroupShuffleSplit(
    n_splits=1,
    test_size=0.20,
    random_state=42
)

train_idx, test_idx = next(
    splitter.split(
        model_data,
        model_data["is_declining_label"],
        groups=model_data["client_id"]
    )
)

train_data = model_data.iloc[train_idx]
test_data = model_data.iloc[test_idx]

# Train the validated Decision Tree
model = DecisionTreeClassifier(
    max_depth=3,
    class_weight="balanced",
    random_state=42
)

model.fit(
    train_data[features],
    train_data["is_declining_label"]
)

# Score all eligible pages for ranking
model_data["model_score"] = model.predict_proba(
    model_data[features]
)[:, 1]

# Rank highest-risk pages first
ranked_queue = model_data.sort_values(
    "model_score",
    ascending=False
).reset_index(drop=True)

ranked_queue["rank"] = ranked_queue.index + 1

# Reason-code function
def assign_reason_code(row):
    reasons = []

    if row["impressions_90d"] <= 5.5:
        reasons.append("LOW_VISIBILITY")

    if row["content_age_days"] > 312.5:
        reasons.append("AGING_CONTENT")

    if row["ctr"] <= 0.30:
        reasons.append("LOW_CTR")

    if row["avg_position"] > 24.45:
        reasons.append("LOW_POSITION")

    if len(reasons) >= 2:
        return "REFRESH_CANDIDATE"

    if len(reasons) == 1:
        return reasons[0]

    return "REVIEW_ONLY"

ranked_queue["reason_code"] = ranked_queue.apply(
    assign_reason_code,
    axis=1
)

# Suggested human-review action
action_map = {
    "LOW_VISIBILITY": "Check search visibility before investing in a refresh.",
    "AGING_CONTENT": "Review whether the content is outdated.",
    "LOW_CTR": "Review title, meta description, and search-intent alignment.",
    "LOW_POSITION": "Review search intent, topical coverage, and internal linking.",
    "REFRESH_CANDIDATE": "Prioritize for a deeper human content audit.",
    "REVIEW_ONLY": "Investigate manually before taking action."
}

ranked_queue["suggested_action"] = ranked_queue["reason_code"].map(
    action_map
)

# Select useful export columns
export_columns = [
    "rank",
    "content_id",
    "client_id",
    "model_score",
    "reason_code",
    "suggested_action"
]

ranked_queue_export = ranked_queue[export_columns].copy()

# Create output directory
output_dir = "/content/Flyrank_Internship_1/work/outputs"
os.makedirs(output_dir, exist_ok=True)

# Export queue
queue_path = os.path.join(
    output_dir,
    "w07_ranked_content_review_queue.csv"
)

ranked_queue_export.to_csv(
    queue_path,
    index=False
)

print("\nQueue exported successfully.")
print("Path:", queue_path)
print("Rows:", len(ranked_queue_export))

print("\nTop 10 ranked pages:")
display(ranked_queue_export.head(10))
```
